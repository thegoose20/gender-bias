Get *Omission* and *Stereotype* label counts by description.

```python
total_descs = 11888
```

```python
df_o = df.loc[df.label == "Omission"]
df_s = df.loc[df.label == "Stereotype"]
print(df_o.shape, df_s.shape)
```

```python
df_o = df_o[["label", "category", "description_id"]]
df_o = df_o.drop_duplicates()
df_s = df_s[["label", "category", "description_id"]]
df_s = df_s.drop_duplicates()
print("Descriptions annotated as Omission:", df_o.shape[0], "("+str((df_o.shape[0]/total_descs)*100)+"%)")
print("Descriptions annotated as Stereotype:", df_s.shape[0], "("+str((df_s.shape[0]/total_descs)*100)+"%)")
```

```python
df_os = pd.concat([df_o, df_s])
print("Descriptions annotated as Omission or Stereotype:", len(df_os.description_id.unique()), "("+str((df_os.shape[0]/total_descs)*100)+"%)")
print("Descriptions not annotated as Omission nor Stereotype:", total_descs-len(df_os.description_id.unique()), 
      "("+str(
          ((total_descs-len(df_os.description_id.unique()))/total_descs)*100)+"%)"
          )
```
